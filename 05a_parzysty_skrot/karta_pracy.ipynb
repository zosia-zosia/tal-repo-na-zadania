## Zadanie 3. Parzysty skrót

Parzystym skrótem dodatniej liczby całkowitej `n` nazwiemy dodatnią liczbę całkowitą `m`, która powstaje przez usunięcie cyfr nieparzystych z zapisu dziesiętnego liczby `n`. Pozostałe cyfry zachowują swoją kolejność.

Parzysty skrót liczby `n` nie istnieje, gdy jej zapis dziesiętny składa się wyłącznie z cyfr nieparzystych. **W takim przypadku funkcja ma zwrócić 0.** Wartość ta oznacza brak skrótu.

W tym zestawie wszystkie liczby wejściowe składają się wyłącznie z cyfr od 1 do 9. Cyfra 0 nie występuje w żadnej liczbie wejściowej.

Przykład:

- Parzystym skrótem liczby 12546 jest liczba 246.
- Parzystym skrótem liczby 8642 jest liczba 8642.
- Parzysty skrót liczby 13579 nie istnieje — funkcja zwraca 0.


## Zadanie 3.1. (0–3)

Napisz funkcję, która dla dodatniej liczby całkowitej `n`, niezawierającej cyfry 0, zwróci jej parzysty skrót. Jeśli parzysty skrót nie istnieje, funkcja ma zwrócić 0.

**Uwaga:** Twój algorytm może używać wyłącznie zmiennych przechowujących liczby całkowite oraz może operować wyłącznie na liczbach całkowitych. W zapisie możesz wykorzystać tylko operacje arytmetyczne: dodawanie, odejmowanie, mnożenie, dzielenie, dzielenie całkowite, resztę z dzielenia oraz porównywanie liczb, instrukcje sterujące, przypisania do zmiennych lub samodzielnie napisane funkcje, wykorzystujące wyżej wymienione operacje. Zabronione jest używanie funkcji wbudowanych oraz operatorów innych niż wymienione.

**Specyfikacja:**

- Dane: `n` — dodatnia liczba całkowita niezawierająca cyfry 0.
- Wynik: `m` — parzysty skrót liczby `n` albo 0, jeśli taki skrót nie istnieje.


````python
def funkcja(n):
    m = 0
    temp = 1
    while n>0:
        cyfra = n%10
        n = n//10
        if cyfra%2==0:
            if cyfra!=0:
                m = m + cyfra*temp
                temp = temp*10
    return m
````

## Zadanie 3.2. (0–3)

Plik `skrot.txt` zawiera 200 dodatnich liczb całkowitych, mniejszych od 30 000. Każda liczba jest zapisana w osobnym wierszu i żadna nie zawiera cyfry 0. Dla co najmniej jednej z tych liczb nie istnieje parzysty skrót.

Napisz program, który wyznaczy liczbę wszystkich liczb z pliku `skrot.txt`, dla których nie istnieje parzysty skrót, oraz poda największą z nich. Odpowiedź zapisz w pliku `wyniki3_2.txt`: w pierwszym wierszu liczbę takich liczb, a w drugim — największą z nich.

Plik `skrot_przyklad.txt` zawiera 20 liczb spełniających warunki zadania. Dla danych zawartych w pliku `skrot_przyklad.txt` prawidłową odpowiedzią jest:

```text
5
19735
```

W pliku jest pięć liczb, dla których nie istnieje parzysty skrót: 13579, 17531, 19735, 11999 i 11111. Największą z nich jest 19735.


````python
def bez_skrotu(n):
    while n > 0:
        cyfra = n % 10
        if cyfra % 2 == 0:
            return False
        n = n // 10
    return True
def wynik(skrot, wyniki3_2):
    n = 0
    maks = 0
    with open(skrot, "r", encoding="utf-8") as plik_in:
        for L in plik_in:
            L = L.strip()
            if not L:
                continue
            liczba = int(L)
            if bez_skrotu(liczba)==True:
                n += 1
                if liczba > maks:
                    maks = liczba
    with open(wyniki3_2, "w", encoding="utf-8") as plik_out:
        plik_out.write(f"{n}\n{maks}\n")
wynik("skrot.txt", "wyniki3_2.txt")
````

## Zadanie 3.3. (0–4)

Plik `skrot2.txt` zawiera 200 dodatnich liczb całkowitych, mniejszych od 30 000. Każda liczba jest zapisana w osobnym wierszu i żadna nie zawiera cyfry 0. Dla każdej z tych liczb istnieje parzysty skrót.

Napisz program, który wypisze te liczby z pliku `skrot2.txt`, dla których największy wspólny dzielnik liczby i jej parzystego skrótu jest podzielny przez 3 bez reszty.

Odpowiedź zapisz w pliku `wyniki3_3.txt`. W każdym wierszu zapisz jedną liczbę z pliku wejściowego spełniającą warunek. Zachowaj kolejność liczb z pliku `skrot2.txt`.

Plik `skrot2_przyklad.txt` zawiera 20 liczb spełniających warunki dotyczące danych wejściowych. Dla danych zawartych w pliku `skrot2_przyklad.txt` prawidłową odpowiedzią jest:

```text
12456
12345
864
1458
6234
```


````python
def skrot(n):
    wynik = 0
    temp = 1
    while n > 0:
        cyfra = n % 10
        n = n // 10
        if cyfra % 2 == 0:
            wynik = wynik + cyfra * temp
            temp = temp * 10
    return wynik
def nwd(a, b):
    while b > 0:
        reszta = a % b
        a = b
        b = reszta
    return a
def rozwiaz(skrot2, wyniki3_3):
    with open(skrot2, "r", encoding="utf-8") as plik_in, \
         open(wyniki3_3, "w", encoding="utf-8") as plik_out:
        for l in plik_in:
            l = l.strip()
            if not l:
                continue
            liczba = int(l)
            skrt = skrot(liczba)
            dz = nwd(liczba, skrt)
            if dz % 3 == 0:
                plik_out.write(f"{liczba}\n")
rozwiaz("skrot2.txt", "wyniki3_3.txt")
````
